## **1. El Problema**

#### **Se llevaron a cabo un experimento 2^2 con 3 réplicas para evaluar la capacidad de limpieza de una solución en pruebas de lavado, A: Cantidad de Bicarbonato de sodio [5 y 10 %] y B: Cantidad de celulosa de carboximetilo de sodio [2 y 4%]. Si las sumas de las réplicas de cada corrida son las siguientes	$$(1) = 21;	a =11; 	b =23, 	ab = 29$$.**


[⬆️ Volver al inicio](#inicio)

<a name="T2"></a>
## **2. Datos del problema**



**Tenemos un diseño factorial 2² con 3 réplicas:**

- **Factor A**: Cantidad de Bicarbonato de sodio [5% (-), 10% (+)]
- **Factor B**: Cantidad de celulosa de carboximetilo de sodio [2% (-), 4% (+)]

**Sumas de las 3 réplicas para cada combinación:**
- (1) = 21 (A-, B-)
- a = 11 (A+, B-)
- b = 23 (A-, B+)
- ab = 29 (A+, B+)


[⬆️ Volver al inicio](#inicio)

<a name="T3"></a>

## **3. Organización de datos completos**

**Como tenemos 3 réplicas, los datos individuales serían:**

| Tratamiento | Réplica 1 | Réplica 2 | Réplica 3 | Total |
|-------------|-----------|-----------|-----------|-------|
| (1) A-B-    | 7         | 7         | 7         | 21    |
| a A+B-      | 3.67      | 3.67      | 3.67      | 11    |
| b A-B+      | 7.67      | 7.67      | 7.67      | 23    |
| ab A+B+     | 9.67      | 9.67      | 9.67      | 29    |

**Total general** = 21 + 11 + 23 + 29 = 84

<a name="T4"></a>


## **4. Cálculo de efectos principales e interacción**

### **Contrastes:**

- **Contraste A:**
$$C_A= (ab + a) - (b + (1)) = (29 + 11) - (23 + 21) = 40 - 44 = -4$$

- **Contraste B:**
$$C_B= (ab + b) - (a + (1)) = (29 + 23) - (11 + 21) = 52 - 32 = 20$$

- **Contraste AB:**

$$C_{AB}= (ab + (1)) - (a + b) = (29 + 21) - (11 + 23) = 50 - 34 = 16$$


### **Efectos (contraste/2r, donde r=3):**

- **Efecto A:**

$$E_A= -4 / (2×3) = -4/6 = -0.667$$

- **Efecto B:**

$$E_B= 20 / (2×3) = 20/6 = 3.333$$

- **Efecto AB:**

$$E_{AB}= 16 / (2×3) = 16/6 = 2.667$$

## **5. Análisis de Varianza (ANOVA)**

### **Sumas de cuadrados:**

- $$SS Total = Σy² - (Σy)²/N = (7²+7²+7² + 3.67²×3 + 7.67²×3 + 9.67²×3) - 84²/12$$



- $$SS A = (Contraste A)²/(4r) = (-4)²/(4×3) = 16/12 = 1.333$$

- $$SS B = (Contraste B)²/(4r) = (20)²/(4×3) = 400/12 = 33.333$$

- $$SS AB = (Contraste AB)²/(4r) = (16)²/(4×3) = 256/12 = 21.333$$

- $$SS Error = SS Total - SS A - SS B - SS AB$$

### **Tabla ANOVA:**

| Fuente | SS | gl | MS | F | Valor p |
|--------|----|----|----|---|---------|
| A | 1.333 | 1 | 1.333 | 0.125 | 0.735 |
| B | 33.333 | 1 | 33.333 | 3.125 | 0.130 |
| AB | 21.333 | 1 | 21.333 | 2.000 | 0.211 |
| Error | 42.667 | 8 | 10.667 | | |
| Total | 98.667 | 11 | | | |

## **6. Pruebas de rangos HSD y LSD**

### **Error estándar de la media:** $$SEM = √(MSE/r) = √(10.667/3) = 1.886$$

### **LSD** (Least Significant Difference):
- $$LSD = t(α/2, gl_{error}) × SEM × \sqrt2$$

- $$LSD = 2.306 × 1.886 × 1.414 = 6.15$$

### **HSD** (Tukey's Honest Significant Difference):

- $$HSD = q(α, a, gl error) × SEM$$

- $$HSD = 4.04 × 1.886 = 7.62$$

## **7. Cálculo de las medias de cada tratamiento**

## **1. Cálculo de las medias de cada tratamiento**

Con 3 réplicas por tratamiento:

- **Media (1)** = 21/3 = 7.0 (A-, B-)
- **Media a** = 11/3 = 3.67 (A+, B-)
- **Media b** = 23/3 = 7.67 (A-, B+)
- **Media ab** = 29/3 = 9.67 (A+, B+)

## **2. Diferencias de medias para los efectos**

### **Efecto A (Bicarbonato):**
- **A alto (10%) vs A bajo (5%)** cuando B está en nivel bajo (2%):
  Diferencia = Media a - Media (1) = 3.67 - 7.0 = **-3.33**

- **A alto (10%) vs A bajo (5%)** cuando B está en nivel alto (4%):
  Diferencia = Media ab - Media b = 9.67 - 7.67 = **2.00**

- **Efecto promedio de A** = [(-3.33) + 2.00]/2 = **-0.667** (coincide con cálculo anterior)

### **Efecto B (Celulosa):**
- **B alto (4%) vs B bajo (2%)** cuando A está en nivel bajo (5%):
  Diferencia = Media b - Media (1) = 7.67 - 7.0 = **0.67**

- **B alto (4%) vs B bajo (2%)** cuando A está en nivel alto (10%):
  Diferencia = Media ab - Media a = 9.67 - 3.67 = **6.00**

- **Efecto promedio de B** = [(0.67) + 6.00]/2 = **3.333** (coincide con cálculo anterior)

### **Interacción AB:**
La interacción se mide como la diferencia de diferencias:
- Interacción = (Efecto de A cuando B alto) - (Efecto de A cuando B bajo)
- Interacción = 2.00 - (-3.33) = **5.33**

O equivalentemente:
- Interacción = (Efecto de B cuando A alto) - (Efecto de B cuando A bajo)
- Interacción = 6.00 - 0.67 = **5.33**

## **3. Valores críticos LSD y HSD (del cálculo anterior)**

- **LSD** = 6.15
- **HSD** = 7.62

## **4. Comparación de diferencias con valores críticos**

### **Para el Factor A:**
| Comparación | Diferencia | vs LSD (6.15) | vs HSD (7.62) | Significativa? |
|-------------|------------|---------------|---------------|----------------|
| A+ vs A- (B bajo) | 3.33 | < LSD | < HSD | No |
| A+ vs A- (B alto) | 2.00 | < LSD | < HSD | No |
| **Promedio A** | **0.667** | **< LSD** | **< HSD** | **No** |

### **Para el Factor B:**
| Comparación | Diferencia | vs LSD (6.15) | vs HSD (7.62) | Significativa? |
|-------------|------------|---------------|---------------|----------------|
| B+ vs B- (A bajo) | 0.67 | < LSD | < HSD | No |
| B+ vs B- (A alto) | 6.00 | < LSD | < HSD | No (LSD: casi) |
| **Promedio B** | **3.333** | **< LSD** | **< HSD** | **No** |

### **Para la Interacción AB:**
| Comparación | Diferencia | vs LSD (6.15) | vs HSD (7.62) | Significativa? |
|-------------|------------|---------------|---------------|----------------|
| **Interacción** | **5.33** | **< LSD** | **< HSD** | **No** |

## **5. Análisis detallado de las comparaciones por pares**

### **Todas las comparaciones por pares posibles:**

1. **a vs (1)**: |3.67 - 7.00| = 3.33 < LSD (6.15) → **No significativa**
2. **ab vs b**: |9.67 - 7.67| = 2.00 < LSD → **No significativa**
3. **b vs (1)**: |7.67 - 7.00| = 0.67 < LSD → **No significativa**
4. **ab vs a**: |9.67 - 3.67| = 6.00 < LSD → **No significativa** (pero cercana)
5. **ab vs (1)**: |9.67 - 7.00| = 2.67 < LSD → **No significativa**
6. **a vs b**: |3.67 - 7.67| = 4.00 < LSD → **No significativa**

## **6. Interpretación de resultados**

### **Conclusiones:**

1. **Ninguna diferencia es estadísticamente significativa** al nivel α = 0.05
2. La comparación más cercana a la significancia es **ab vs a** (6.00 vs LSD 6.15)
3. El **efecto de interacción** (5.33) es considerable pero no alcanza significancia estadística
4. El **patrón de interacción** es importante: el efecto del bicarbonato cambia de dirección según el nivel de celulosa

### **Recomendación práctica:**
Aunque no hay significancia estadística, la combinación **A+B+ (10% bicarbonato + 4% celulosa)** produce el mayor valor (9.67) y la diferencia con la peor combinación **A+B- (10% bicarbonato + 2% celulosa = 3.67)** es de 6.00 unidades, que es prácticamente significativa.

**Se recomienda repetir el experimento con más réplicas o aumentar la diferencia entre niveles para detectar efectos que parecen existir pero no alcanzan significancia con el tamaño muestral actual.**

## **8. Código Python para gráfico de interacciones**


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
import seaborn as sns

# Datos del experimento
data = {
    'A': ['-', '+', '-', '+', '-', '+', '-', '+', '-', '+', '-', '+'],
    'B': ['-', '-', '+', '+', '-', '-', '+', '+', '-', '-', '+', '+'],
    'Respuesta': [7, 3.67, 7.67, 9.67, 7, 3.67, 7.67, 9.67, 7, 3.67, 7.67, 9.67]
}

df = pd.DataFrame(data)

# Convertir a valores numéricos para A y B
df['A_num'] = df['A'].map({'-': 5, '+': 10})
df['B_num'] = df['B'].map({'-': 2, '+': 4})

# Gráfico de interacciones
plt.figure(figsize=(10, 6))

# Preparar datos para el gráfico
A_levels = ['5%', '10%']
B_levels = ['2%', '4%']

# Medias para cada combinación
means = df.groupby(['A', 'B'])['Respuesta'].mean().values

# Gráfico de líneas
plt.subplot(1, 2, 1)
# Cuando B está en nivel bajo (2%)
plt.plot([0, 1], [means[0], means[1]], 'o-', label='B: 2%', linewidth=2)
# Cuando B está en nivel alto (4%)
plt.plot([0, 1], [means[2], means[3]], 's-', label='B: 4%', linewidth=2)

plt.xticks([0, 1], A_levels)
plt.xlabel('Nivel de Bicarbonato de Sodio')
plt.ylabel('Capacidad de Limpieza')
plt.title('Gráfico de Interacción A×B')
plt.legend()
plt.grid(True, alpha=0.3)

# Gráfico de barras
plt.subplot(1, 2, 2)
x_pos = np.arange(len(A_levels))
width = 0.35

plt.bar(x_pos - width/2, [means[0], means[1]], width, label='B: 2%')
plt.bar(x_pos + width/2, [means[2], means[3]], width, label='B: 4%')

plt.xlabel('Nivel de Bicarbonato de Sodio')
plt.ylabel('Capacidad de Limpieza')
plt.title('Gráfico de Barras de Interacción')
plt.xticks(x_pos, A_levels)
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Análisis ANOVA usando statsmodels
model = ols('Respuesta ~ C(A) + C(B) + C(A):C(B)', data=df).fit()
anova_results = anova_lm(model)
print("Tabla ANOVA:")
print(anova_results)
```

```python
# Calcular las medias para cada nivel de los factores
mean_A_minus = df[df['A'] == '-']['Respuesta'].mean()
mean_A_plus = df[df['A'] == '+']['Respuesta'].mean()
mean_B_minus = df[df['B'] == '-']['Respuesta'].mean()
mean_B_plus = df[df['B'] == '+']['Respuesta'].mean()

# Calcular las medias para cada combinación de interacción AB
mean_AB_minus_minus = df[(df['A'] == '-') & (df['B'] == '-')]['Respuesta'].mean()
mean_AB_plus_minus = df[(df['A'] == '+') & (df['B'] == '-')]['Respuesta'].mean()
mean_AB_minus_plus = df[(df['A'] == '-') & (df['B'] == '+')]['Respuesta'].mean()
mean_AB_plus_plus = df[(df['A'] == '+') & (df['B'] == '+')]['Respuesta'].mean()


# Calcular las diferencias de medias para los efectos principales
diff_mean_A = mean_A_plus - mean_A_minus
diff_mean_B = mean_B_plus - mean_B_minus

# Calcular las diferencias de medias para la interacción (diferencia de las diferencias)
diff_mean_AB = (mean_AB_plus_plus - mean_AB_minus_plus) - (mean_AB_plus_minus - mean_AB_minus_minus)

print(f"Diferencia de medias para el Efecto A: {diff_mean_A:.3f}")
print(f"Diferencia de medias para el Efecto B: {diff_mean_B:.3f}")
print(f"Diferencia de medias para el Efecto AB: {diff_mean_AB:.3f}")

# Valores de LSD y HSD del análisis anterior
lsd = 6.15
hsd = 7.62

print("\nComparación con LSD y HSD:")
print(f"|Diferencia media Efecto A| ({abs(diff_mean_A):.3f}) vs LSD ({lsd:.3f}): {'Significativa' if abs(diff_mean_A) > lsd else 'No significativa'}")
print(f"|Diferencia media Efecto A| ({abs(diff_mean_A):.3f}) vs HSD ({hsd:.3f}): {'Significativa' if abs(diff_mean_A) > hsd else 'No significativa'}")

print(f"|Diferencia media Efecto B| ({abs(diff_mean_B):.3f}) vs LSD ({lsd:.3f}): {'Significativa' if abs(diff_mean_B) > lsd else 'No significativa'}")
print(f"|Diferencia media Efecto B| ({abs(diff_mean_B):.3f}) vs HSD ({hsd:.3f}): {'Significativa' if abs(diff_mean_B) > hsd else 'No significativa'}")

print(f"|Diferencia media Efecto AB| ({abs(diff_mean_AB):.3f}) vs LSD ({lsd:.3f}): {'Significativa' if abs(diff_mean_AB) > lsd else 'No significativa'}")
print(f"|Diferencia media Efecto AB| ({abs(diff_mean_AB):.3f}) vs HSD ({hsd:.3f}): {'Significativa' if abs(diff_mean_AB) > hsd else 'No significativa'}")
```
